Import libraries

```python
import pandas as pd
```

process lists

```python
def add_column(tableID):
    editorDf = pd.read_csv(f'https://docs.google.com/spreadsheets/d/1qSSC-39SewS3ifjeHl7_lUdgN34c-vVzInJuhkHC0ig/export?format=csv&gid={tableID}', header=None)
    # print(editorDf)
    connectionDf = pd.read_csv(f'https://docs.google.com/spreadsheets/d/15mjhEAJDR5rhdux3mdjVVLWATm20uo_droIn5K4kRXA/export?format=csv&gid=1532912149')
    # print(connectionDf)
    results = []
    for index, row in connectionDf.iterrows():
        myFromId = row['from']
        myToId = row['to']
        mask = (editorDf.iloc[:, 0] == myFromId) & (editorDf.iloc[:, 6] == myToId)
        if mask.any():
            row_index = editorDf.index[mask][0]             # find the matching row index
            connection = editorDf.iloc[row_index, 12]       # column 13 (index 12)
            results.append(connection)
        else: 
            results.append('')
            
    # turn results into a DataFrame
    resultDf = pd.DataFrame(results)

    # copy to clipboard for Google Sheets
    resultDf.to_clipboard(excel=True, index=False, header=False)
```

execute function

```python
add_column('73087843')
```
