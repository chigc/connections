Import libraries

```python
from lxml import html, etree
import json, pyperclip
```

Declare namespaces

```python
ns = {'svg': 'http://www.w3.org/2000/svg'}
```

Process connections table with navgraph svg in connections html

```python
myHTML = html.parse('connections.html', html.HTMLParser(remove_blank_text=True))
nodeGroup = myHTML.xpath('//g[@id="nodes"]')[0]
myJSON = json.load(open('nodes.json'))
# print(myJSON)
for node in nodeGroup.findall('g', ns):
    nodeID = node.get('id')
    classes = []
    for key, values in myJSON.items():
        if nodeID in values:
            classes.append(key)
    # print(f'Node {nodeID} belongs to classes {classes}')
    node.set('class', ' '.join(classes))

outputGroup = etree.tostring(nodeGroup, encoding='unicode', pretty_print=True)
print(outputGroup)
pyperclip.copy(outputGroup) #copy to clipboard
```
